# DAGMC Model & Mesh Inspection

This workbook shows the commands for loading step and dagmc models for viewing in pyvista.

Geometry generated with code from Paramak's `tokamak_with_pf_magnets.py`:

(https://github.com/fusion-energy/paramak/blob/main/examples/tokamak_with_pf_magnets.py)  

```python
import stellarvista as sv
```

```python
# After importing the model, it behaves as a normal pv.MultiBlock.
dagmc_model = sv.import_dagmc('./model_paramak/dagmc.h5m')
display(dagmc_model) # display the model to view the DAGMC cells contained within

# Create a STATIC image
dagmc_model.plot(show_edges=True, jupyter_backend='static', cpos=[4.0, -4.0, 1.0])

# PyVista makes it very easy to clip and apply other model transformations for easy inspection.
# This will create an interactive visualization.
dagmc_model.clip('x').plot(show_edges=True, cpos=[4.0, -4.0, 1.0])
```

```python
# We can also isolate individual cell(s) for inspection
component = dagmc_model['mat_10']
display(component)
component.plot(show_edges=True, cpos=[4.0, -4.0, 1.0], color='gray')
```
