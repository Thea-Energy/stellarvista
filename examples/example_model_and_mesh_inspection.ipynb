This workbook shows the commands for loading step and dagmc models for viewing in pyvista.

Geometry generated with code from Paramak's tokamak_with_pf_magnets.py example:
    https://github.com/fusion-energy/paramak/blob/main/examples/tokamak_with_pf_magnets.py

```python
import stellarvista as sv
```

```python
model = sv.import_step('./model_paramak/tokamak_model.step')
display(model[0])
model.plot(scalars='SUBSHAPE_IDS',show_edges=True, jupyter_backend='static',  cpos=[4.0, -4.0, 1.0])
```

```python
dagmc_model = sv.import_h5m('./model_paramak/dagmc.h5m')
display(dagmc_model)
dagmc_model.plot(show_edges=True, jupyter_backend='static', cpos=[4.0, -4.0, 1.0])
dagmc_model.clip('x').plot(show_edges=True, jupyter_backend='static', cpos=[4.0, -4.0, 1.0])
```

```python
component = dagmc_model['mat_10']
display(component)
component.plot(show_edges=True, jupyter_backend='static', cpos=[4.0, -4.0, 1.0], color='gray')
```
